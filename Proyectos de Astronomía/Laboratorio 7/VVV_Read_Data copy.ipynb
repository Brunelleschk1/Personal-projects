```python
#Para leer archivos HDF5, estilo base de datos jerárquica, pueden usar el siguiente código:

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import h5py
```

```python
def phase(t,f):
    return np.mod(t*f,1)

db = "singlemode_variable_stars_sample.h5"

with h5py.File(db,"r") as file:
    star_codes = list(file["/"].keys())

# para abrir la estrella número 15
name = star_codes[15]
with h5py.File(db,"r") as file:
    star_data = pd.DataFrame(file[name][:])

plt.scatter(star_data.RHJD, star_data.mag,c='k',s=1)
plt.errorbar(star_data.RHJD, star_data.mag, star_data.mag_err, fmt=" k", capsize=3, alpha=0.1)
plt.gca().invert_yaxis()
plt.title(f"star with code: {name}")
plt.xlabel("Time (RHJD)")
plt.ylabel("I magnitude")
```
